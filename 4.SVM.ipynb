# Support Vector Machine (SVM)

**How can we decide between perfect classifiers?**
<center><img src="img/fig401.png" width=400 height=400 /></center>

- **Define the margin to be the distance of the closest data point to the classifier**
<center><img src="img/fig40.png" width=400 height=400 /></center>



- **A new example not from the training set might be misclassified if the margin is smaller**
<center><img src="img/fig41.png" width=400 height=300 /></center>

- **Support vector machines (SVMs):** Choose the classifier with the largest margin
    - Has good practical and theoretical performance

## Some Geometry

<center><img src="img/fig42.png" width=500 height=400 /></center>


**The distance of any point to a hyperplane is given by**

$$r = \frac{y_{i} (w^{T} \cdot x_{i} + b)}{||w||} = \frac{y_{i} f(x_{i})}{||w||}$$

- This means the distance from the origin $x=0$
<center>$$r = \frac{f(0)}{||w||} = \frac{w \cdot 0 + b }{||w||} = \frac{b }{||w||} $$</center>

- **To obtain a unique hyperplane**, we choose a point where its distance from the hyperplane is $1$, $y^{*}f(x^{*}) = 1$ so **the margin is** given as
<center>$$m = \frac{y^{*} (w^{T} \cdot x^{*} + b)}{||w||} = \frac{1}{||w||}$$</center>

- We call these points **support vectors**


- We want to maximize the margin subject to the constraints that

$$y_{i}(w \cdot x_{i} + b) >= 1$$

### The optimization problem of SVM (Hard SVM)
$$\large max_{w,b} \frac{1}{||w||}$$
$$ \large \text{s.t.} \forall i, \space y_{i}(w \cdot x_{i} + b) >= 1  $$

**Or, equivalently,**

$$\large min_{w,b} ||w||^{2}$$
such that, $$ \large \text{s.t.} \forall i,\space y_{i}(w \cdot x_{i} + b) >= 1$$

    - This is a standard quadratic programming problem
    - Falls into the class of convex optimization problems
    - Can be solved with many specialized optimization tools 

### What if the data is not separable?

- **For hard SVM**, the problem will be infeasible, no solution

### Soft SVM

**The idea: Allow some examples to “break into the margin"**

<center><img src="img/fig44.png" width=400 height=400 /></center>

- Introduce **one slack variable** $\zeta_{i}$ per example
- And require $\space$ $y_{i} w^{T} x_{i} >= 1- \zeta_{i}$ $\space$ and $\space$ $\zeta_{i} >= 0$ 

#### New optimization problem for learning

<center><img src="img/fig45.png" width=300 height=300 /></center>

- Minimize $||w||^{2}$ means maximize the margin **($L2$ norm regularization)**
- Minimize total slack (or the loss) means allow as few examples as possible to violate the margin
- The cost $C$ trade off between the two terms

### Primal SVM

#### Substitute the slack variables to reformulate the constrained soft-margin SVM into an unconstrained optimization

<center><img src="img/fig46.png" width=400 height=300 /></center>

**We can consider three cases**
- Example is correctly classified and is outside the margin: $\text{penalty} = 0$
- Example is incorrectly classified: $\text{penalty} = 1 - y_{i} w^{T} x_{i}$
- Example is correctly classified but within the margin: $\text{penalty} = 1 - y_{i} w^{T} x_{i}$

**This is the hinge loss function**

## Dual SVM Formulation

- #### Using Lagrange Multipliers

Use Lagrange multipliers $\alpha_{i}$ to include the constraints in the optimization problem. 

<center><img src="img/fig54.png" width=350 height=350 /></center>


We turn this into dual Lagrangian by taking the gradient with respect to $w$ and $b$, then substituting these results into $L$ 

$$ \sum_{i=1}^n \alpha_n   -\frac{1}{2} \sum_{i=1}^n \sum_{j=1}^n \alpha_i \alpha_j y_i y_j \phi(x_{i})^T \phi(x_j)$$

such that $\alpha_i \ge 0$ and $\sum_{i=1}^n \alpha_i y_i = 0$.



- Once we have the solution for $\alpha_{i}$, all $\alpha_{i} > 0 $ are support vectors
- The discriminant function is given by

$$\delta(x) = sign(w^{T}x + b)$$  

Where we can obtain $w$ and $b$ from the support vectors

$$w = \sum\limits_{\alpha_{i} > 0} \alpha_i y_i x_i$$
$$b = avg_{\alpha_{i} > 0} (\frac{1}{y_{i}} - w^{T} x_{i})$$

## Kernel trick

- In order to find a halfspace predictor in the high dimensional space, we do not need to know the representation of instances in that space,
- but rather the values of inner products between the mapped instances.


<center><img src="img/fig53.png" width=300 height=300 /></center>

### Gaussian Kernel (RBF)

$$K(x,x') = exp(-\frac{||x-x'||}{2\sigma}) $$

### Polynomial Kernel

$$K(x,x') = (c + \gamma (x,x'))^{d}$$

```python
def gaussian_kernel(xi, xj):
    return np.exp(-np.sum((xi - xj)**2) / 20.0)

def poly_kernel(xi, xj):
    degree = 4
    return (1 + np.dot(xi, xj)) ** degree
    
N = X.shape[0]
gK = np.zeros((N,N))
pK = np.zeros((N,N))
for i in range(N):
    for j in range(N):
        gK[i, j] = gaussian_kernel(X[i, :], X[j, :])
        pK[i, j] = poly_kernel(X[i, :], X[j, :])

```

### Multiclass Classification

**- One-Versus-All SVMs**

    - Compute a classifier for each label versus the remaining labels
    - If there are 𝐿 possible labels, requires learning 𝐿 classifiers over the entire data set

<center><img src="img/fig43.png" width=300 height=300 /></center>

**- One-Versus-One SVMs**

    - Construct a classifier for all possible pairs of labels
    - Given a new data point, can classify it by majority vote
- If there are 𝐿 labels, requires computing $\binom{L}{2}$ different classifiers

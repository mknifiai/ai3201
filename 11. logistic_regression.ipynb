## Logistic Regression

**The setting**
    
- Binary classification
- Inputs: Feature vectors $x \in R^d$
- Labels: $y \in \{−1, +1\}$

**Training data**
    
- $S = \{(x_i, y_i)\}$, consisting of $m$ examples

Instead of predicting a label, we will predict $P(y = +1 ∣ x)$

Expand hypothesis space to functions whose output is in the
range $[0, 1]$

- Original problem: $R^d → \{−1, +1\}$
- Modified problem: $R^d → [0, 1]$
- Effectively, make the problem a regression problem

### The Sigmoid function

The hypothesis space for logistic regression: All functions of the form

$$h_{\mathbf{w}}(\mathbf{x}) = \sigma(\mathbf{w}^T\mathbf{x}) = \frac{1}{1 + \exp(-\mathbf{w}^T\mathbf{x})}
$$

That is, a linear function, composed with a sigmoid function (the **logistic function**), defined as

$$\sigma(z) = \frac{1}{1 + \exp(-z)} = \frac{\exp(z)}{1 + \exp(z)}$$

What is its derivative with respect to z?

$$\sigma(z) (1 - \sigma(z))$$

**What is the domain
and the range of the
sigmoid function?**

<center><img src="img/fig79.png" width=300 height=300 /></center>

- The **logistic function** “squashes” the output to be between 0 and 1 for any scalar input $z$. 

- The output $\theta(z)$ can therefore be interpreted as a probability.


### Predicting probabilities

According to the logistic regression model, we have

$$P(y = 1 \mid \mathbf{x}; \mathbf{w}) = \sigma(\mathbf{w}^T\mathbf{x}) = \frac{1}{1 + \exp(-\mathbf{w}^T\mathbf{x})}$$

$$P(y = -1 \mid \mathbf{x}; \mathbf{w}) = 1 - \sigma(\mathbf{w}^T\mathbf{x}) = \frac{1}{1 + \exp(\mathbf{w}^T\mathbf{x})}$$


Or equivalently

$$P(y \mid \mathbf{x}; \mathbf{w}) = \frac{1}{1 + \exp(-y\mathbf{w}^T\mathbf{x})}
$$

**Note that we are directly modeling
P(y| x) rather than P(x |y) and P(y)**

### Training a logistic regression classifier

– **First:** Maximum likelihood estimation

– **Then:** Adding priors → Maximum a Posteriori estimation

#### Maximum likelihood estimation

Find a weight vector $w$ such that $P(S ∣ w)$ is maximized
    
    - Maximum likelihood training of a discriminative probabilistic classifier under the logistic model for the posterior distribution. 

$$\underset{\mathbf{w}}{\operatorname{argmax}} \, P(S \mid \mathbf{w}) = \underset{\mathbf{w}}{\operatorname{argmax}} \prod_{i=1}^{m} P(y_i \mid \mathbf{x}_i, \mathbf{w}) $$
- The trick: Convert products to sums by taking log
    - this works only because log is an increasing
    - function and the maximizer will not change


$$\max_{\mathbf{w}} \sum_{i}^{m} \log P(y_i \mid \mathbf{x}_i, \mathbf{w})$$

But (by definition) we know that

$$P(y_i \mid \mathbf{w}, \mathbf{x}_i) = \sigma(y_i \mathbf{w}^T \mathbf{x}_i) = \frac{1}{1 + \exp(-y_i \mathbf{w}^T \mathbf{x}_i)}$$

So it is equivalent to solving

$$\max_{\mathbf{w}} \sum_{i}^{m} -\log(1 + \exp(-y_i \mathbf{w}^T \mathbf{x}_i))$$

or equivalently

$$\min_{\mathbf{w}} \sum_{i}^{m} \log(1 + \exp(-y_i \mathbf{w}^T \mathbf{x}_i))$$
    
    - This is a training of a linear classifier by minimizing the logistic loss. 



#### Maximum a posteriori estimation

**What is the goal of MAP estimation?**

- To maximize the posterior probability of the model given the data (i.e. to find the
most probable model, given the data)
$$p(w|𝑆) = P(𝑆|w) P(w)$$

To maximize

$$\underset{\mathbf{w}}{\operatorname{argmax}} \, P(\mathbf{w} \mid S) = \underset{\mathbf{w}}{\operatorname{argmax}} \, P(S \mid \mathbf{w}) P(\mathbf{w})$$

Take log to simplify

$$\max_{\mathbf{w}} \log P(S \mid \mathbf{w}) + \log P(\mathbf{w})$$

- In maximum likelihood estimation, we maximized the likelihood of the data ($\max_{\mathbf{w}} \log P(S \mid \mathbf{w})$)

- Regrading the prior $P(w)$:

    - we assume weight in the weight vector is drawn independently from the normal
      distribution with zero mean and standard deviation $\sigma$

$$p(\mathbf{w}) = \prod_{j=1}^{d} p(w_i) = \prod_{j=1}^{d} \frac{1}{\sigma\sqrt{2\pi}} \exp\left( \frac{-w_i^2}{\sigma^2} \right)
$$

**Learning a logistic regression classifier is equivalent to
solving**

$$\min_{\mathbf{w}} \sum_{i}^{m} \log(1 + \exp(-y_i \mathbf{w}^T \mathbf{x}_i)) + \frac{1}{\sigma^2} \mathbf{w}^T \mathbf{w}
$$

We can solve this function for $w$ by 
- the LBFGS algorithm (a quasi-Newton method).
- But gradient based methods like SGD and its variants are way more commonly used.

## Empirical loss minimization

$\textbf{Learning = minimize empirical loss on the training set}$

$$
\min_{h \in H} \frac{1}{m} \sum_{i} L(h(\mathbf{x}_i), f(\mathbf{x}_i))
$$


**Is there a problem here?** $\textcolor{red}{Overfitting!}$


- We need something that biases the learner towards simpler hypotheses
    - Achieved using a $\textcolor{blue}{regularizer}$, which penalizes complex hypotheses



### Regularized loss minimization

- using $\ell_2$ regularization
$$\min_{\mathbf{w}} \frac{1}{2}\mathbf{w}^T\mathbf{w} + C \sum_{i} L(y_i, \mathbf{x}_i ,\mathbf{w})$$

**What is a loss function?**

- Loss functions should penalize mistakes
- We are minimizing average loss over the training data


#### What is the ideal loss function for classification?

Penalize classification mistakes between true label $y$ and prediction $y'$

$$
L_{0\text{-}1}(y, y') = 
\begin{cases} 
1 & \text{if } y \neq y', \\ 
0 & \text{if } y = y'. 
\end{cases}
$$

* For linear classifiers, the prediction $y' = \operatorname{sgn}(\mathbf{w}^T\mathbf{x})$
    * Mistake if $y \mathbf{w}^T\mathbf{x} \le 0$

$$
L_{0\text{-}1}(y, \mathbf{x}, \mathbf{w}) = 
\begin{cases} 
1 & \text{if } y \mathbf{w}^T\mathbf{x} \le 0, \\ 
0 & \text{otherwise}. 
\end{cases}
$$

### The loss function zoo

#### Many loss functions exist

* **Perceptron loss**
  $$L_{Perceptron}(y, \mathbf{x}, \mathbf{w}) = \max(0, -y\mathbf{w}^T\mathbf{x})$$

* **Hinge loss (SVM)**
  $$L_{Hinge}(y, \mathbf{x}, \mathbf{w}) = \max(0, 1 - y\mathbf{w}^T\mathbf{x})$$

* **Exponential loss (AdaBoost)**
  $$L_{Exponential}(y, \mathbf{x}, \mathbf{w}) = e^{-y\mathbf{w}^T\mathbf{x}}$$

* **Logistic loss (logistic regression)**
  $$L_{Logistic}(y, \mathbf{x}, \mathbf{w}) = \log(1 + e^{-y\mathbf{w}^T\mathbf{x}})$$

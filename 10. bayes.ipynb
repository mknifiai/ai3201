## 1. Bayes Classifier

**The setting**
- Classification task
- Inputs: Training data $X \in \mathbf{R}^{n\text{x}d}$
- Labels: $Y \in \{c_1, c_2, \dots, c_k\}$

The Bayes classifier uses the Bayes theorem to predict the class for a new test instance, $\mathbf{x}$. 

It estimates **the posterior probability** $P(c_i | \mathbf{x})$ for each class $c_i$, and chooses the class that has the largest probability. 

The predicted class for $\mathbf{x}$ is given as
\begin{equation*}
\hat{y} = \underset{c_i}{\arg\max} \{P(c_i | \mathbf{x})\} 
\end{equation*}

**The Bayes theorem** allows us to invert the posterior probability in terms of the likelihood and prior probability, as follows:
\begin{equation*}
P(c_i | \mathbf{x}) = \frac{P(\mathbf{x} | c_i) \cdot P(c_i)}{P(\mathbf{x})} 
\end{equation*}

- where $P(\mathbf{x} | c_i)$ is the likelihood, defined as the probability of observing $\mathbf{x}$ assuming that the true class is $c_i$
- $P(c_i)$ is the prior probability of class $c_i$
- $P(\mathbf{x})$ is the probability of observing $\mathbf{x}$ from any of the $k$ classes, given as
\begin{equation*}
P(\mathbf{x}) = \sum_{j=1}^{k} P(\mathbf{x} | c_j) \cdot P(c_j)
\end{equation*}

Because $P(\mathbf{x})$ is fixed, the Bayes rule can be rewritten as:

\begin{equation*}
\hat{y} = \underset{c_i}{\arg\max} \{ P(c_i | \mathbf{x}) \} = \underset{c_i}{\arg\max} \left\{ \frac{P(\mathbf{x} | c_i) P(c_i)}{P(\mathbf{x})} \right\} \nonumber \\
= \underset{c_i}{\arg\max} \{ P(\mathbf{x} | c_i) P(c_i) \} 
\end{equation*}

- the predicted class depends on the likelihood of that class taking its prior probability into account.



### Estimating the Prior Probability

The prior probability for class $c_i$ can be estimated as follows:
\begin{equation*}
\hat{P}(c_i) = \frac{n_i}{n}
\end{equation*}

where $n$ is the size of the dataset and $n_i$ is the size of $i^{th}$ class. 

### Estimating the Likelihood

To estimate the likelihood $P(\mathbf{x}|c_i)$, we have to estimate the joint probability of $\mathbf{x}$ across all the $d$ dimensions, 
$$P(\mathbf{x} = (x_1, x_2, \dots, x_d)|c_i)$$


For numeric attribute, the joint probability of $\mathbf{x}$ can be estimated via either a

- nonparametric
- or a parametric approach.


**In the parametric approach**
- assume that each class $c_i$ is normally distributed around some mean $\boldsymbol{\mu}_i$ with a corresponding covariance matrix $\boldsymbol{\Sigma}_i$, both of which are estimated from $\mathbf{D}_i$. 

For class $c_i$, the probability density at $\mathbf{x}$ is thus given as
\begin{equation*}
f_i(\mathbf{x}) = f(\mathbf{x}|\boldsymbol{\mu}_i, \boldsymbol{\Sigma}_i) = \frac{1}{(\sqrt{2\pi})^d \sqrt{|\boldsymbol{\Sigma}_i|}} \exp \left\{ -\frac{(\mathbf{x} - \boldsymbol{\mu}_i)^T \boldsymbol{\Sigma}_i^{-1} (\mathbf{x} - \boldsymbol{\mu}_i)}{2} \right\} 
\end{equation*}

We can predict the class for $\mathbf{x}$ as follows:

$$    \hat{y} = \arg\max_{c_i} \left\{ f_i(\mathbf{x}) P(c_i) \right\}$$


To classify a numeric test point $\mathbf{x}$, the Bayes classifier estimates 
the parameters via the sample mean and sample covariance matrix. 

- The sample mean for the class $c_i$ can be estimated as
$$
    \hat{\boldsymbol{\mu}}_i = \frac{1}{n_i} \sum_{\mathbf{x}_j \in D_i} \mathbf{x}_j
$$

- And the sample covariance matrix for each class can be estimated using as follows
$$
    \hat{\boldsymbol{\Sigma}}_i = \frac{1}{n_i} \overline{\mathbf{D}}_i^T \overline{\mathbf{D}}_i
$$

where $\overline{\mathbf{D}}_i$ is the centered data matrix for class $c_i$ 
given as $\overline{\mathbf{D}}_i = \mathbf{D}_i - \hat{\boldsymbol{\mu}}_i^T$. 

>> covariance matrix is a square matrix summarizing how a set of numeric variables vary—both individually and in relation to one another across multiple dimensions


## 2. Naive Bayes Classifier

The naive Bayes approach assumes that all the attributes are **independent**. 

- The independence assumption implies that the likelihood can be decomposed into a product 
of dimension-wise probabilities:

$$
P(\mathbf{x}|c_i) = P(x_1, x_2, \ldots, x_d|c_i) = \prod_{j=1}^{d} P(x_j|c_i)
$$

We assume that each attribute is normally distributed for each class $c_i$.

Let $\mu_{ij}$ and $\sigma_{ij}^2$ denote the mean and variance for attribute $X_j$, for class $c_i$.

The likelihood for class $c_i$, for dimension $X_j$, is given as

$$
P(x_j|c_i) \propto f(x_j|\mu_{ij}, \sigma_{ij}^2) = \frac{1}{\sqrt{2\pi}\sigma_{ij}} \exp \left\{ -\frac{(x_j - \mu_{ij})^2}{2\sigma_{ij}^2} \right\}
$$

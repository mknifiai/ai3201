#### Linear Separators

- In $d$ dimensions, a hyperplane is a solution to the equation

<center>$w^{T}x + b = 0$</center>

with $w \in R^{d}$ and $b \in R$

- **Hyperplanes** divide $R^{d}$ into two distinct sets of points (called open **halfspace**)

<center>$w^{T}x + b > 0$</center>

<center>$w^{T}x + b < 0$</center>

- Hypothesis space: separating hyperplanes

<center>$f(x) = sign(w^{T}x + b)$</center>

- How should we choose the loss function?

    - Count the number of misclassifications 
        - $loss = \sum\limits_{i=1}^{N} |y_{i} - sign(w^{T}x_{i} + b)|$
        - Tough to optimize, gradient contains no information (Flat everywhere except at zero)

    - Penalize misclassification linearly by the size of the violation
        - $perceptron$ $loss = \sum\limits_{i=1}^{N} max\{0,- y_{i} (w^{T}x_{i} + b)\}$
        - $perceptron$ $loss$ is convex but not differentiable everywhere (The solution is to use **subgradient**)

#### Subgradient

- For a convex function $g(x)$, a subgradient at a point $x_{0}$ is given
- by any line, $l$, such that $l(x_{0}) = g(x_{0})$ and $l(x) \leq g(x)$ for all
$x$,
- i.e., it is a linear underestimator   

<center><img src="img/fig34.png" width=300 height=300 /></center>

- If a convex function is **differentiable** at a point $x$, then it has a
unique subgradient at the point $x$ given by the gradient

- If a convex function is **not differentiable** at a point $x$, it can have
many subgradients

    • E.g., the set of subgradients of the convex function $|x|$ at the
  point $x=0$ is given by the set of slopes $[−1,1]$

- **Subgradients** only guaranteed to exist for convex functions

        - Try to minimize the perceptron loss using (sub)gradient descent

$w_{t+1} = w_{t} + \gamma_{t} \sum\limits y_{i}x_{i} . 1_{-y_{i} f_{w,b}(x_{i}) \geq 0} $ 

$b_{t+1} = b_{t} + \gamma_{t} \sum\limits y_{i} . 1_{-y_{i} f_{w,b}(x_{i}) \geq 0} $

>> the indicator function is equal to zero if the $x_{i}$ is correctly classified and one otherwise)

where $\gamma$ is the step size (or learning rate). The step size must gradually decrease toward zero $\gamma = \frac{1}{1+t}$

- Stochastic gradient descent converges to the global optimum
under certain assumptions on the step size

### 2. Stochastic Gradient Descent Algorithm

- pick a random observation $m$ and perform the following update

- if the $m^{th}$ data point is misclassified:

<center>$$w_{t+1} = w_{t} + \gamma_{t} y_{m} x_{m}$$</center>
    

<center>$$b_{t+1} = b_{t} + \gamma_{t} y_{m}$$</center>

- if the $m^{th}$ data point is correctly classified:

<center>$$w_{t+1} = w_{t}$$</center>

<center>$$b_{t+1} = b_{t}$$</center>

- Perceptron algorithm sometimes specified with $\gamma_{t} = 1$ for all $t$

### 3. Perceptron Learning Drawbacks

- No convergence guarantees if the observations are not linearly separable

- Can overfit

    - There can be a number of perfect classifiers,
    - but the perceptron algorithm doesn’t have any mechanism for choosing between them

### 5. What If the Data Isn‘t Separable?

- e.g., The XOR function

<center><center><img src="img/fig36.png" width=300 height=300 /></center></center>

- Can add features to make the data linearly separable in a higher dimensional space!
    - Choose a feature map $\Phi: R^{d}\rightarrow R^{k}$
    - Given data $x_{1}\dots x_{n}$, we can construct feature vectors $\Phi(x_{1})\dots \Phi(x_{n})$
    - Use $\Phi(x)$ instead of $(x)$ in the learning algorithm
    - Learn linear separators of the form $w\cdot\Phi(x)$ instead of $w\cdot x$

Say we have a single feature for each sample.  Our data matrix is
$$
\begin{alignat*}{1}
 X &= \begin{bmatrix}
       x_0\\
       x_1\\
       \vdots \\
       x_N
       \end{bmatrix}
\end{alignat*}
$$
We can add other powers of each $x$ value, say up to the fourth power.
$$
\begin{alignat*}{1}
 \Phi(X) &= \begin{bmatrix}
       x_0 & x_0^2 & x_0^3 & x_0^4\\
       x_1 & x_1^2 & x_1^3 & x_1^4\\
       \vdots \\
       x_N & x_N^2 & x_N^3 & x_N^4\\
       \end{bmatrix}
\end{alignat*}
$$

- **Which of these feature are useful?**
    - Looking at the magnitudes of the weights is helpful, as long as **the input features are standardized.**

- **Notice** that more expressive features can lead to **overfitting!**

### 6. Why Feature scaling or Standardization Matters?

- **Standardization** rescales each feature so it has a mean of 0 and a standard deviation of 1

<center>$$x' = \frac{x - \mu}{\sigma}$$</center>

    - It's what makes weight magnitudes comparable 

- **Normalization** (Min-Max scaling to [0, 1])
    - Normalization most commonly refers to min-max scaling, which squeezes values into a fixed range, typically [0,1]
<center>$$x' = \frac{x - x_{min}}{x_{max} - x_{min}}$$</center>

- **Scaling to [-1, 1]**
  <center>$$x' = 2. \frac{x - x_{min}}{x_{max} - x_{min}} - 1$$</center>


- $L^2$ **normalization**
    - It is a different operation applied per-row rather than per-column

### 7. Classification Assessment

- #### Confusion Matrix
<center><center><img src="img/fig38.png" width=500 height=300 /></center></center>

**Metrics Derived from Confusion Matrix**

- **Accuracy**
    -  measures the overall fraction of predictions that were correct.
    -  It works well when classes are balanced but can be misleading otherwise.

<center>$$ Accuracy = \frac{TP+TN}{TP+TN+FP+FN}$$</center>


- **Error Rate (Misclassification Rate)**

$$Error Rate = 1 − Accuracy$$


- **Precision** (also called Positive Predictive Value)
    - tells you what proportion of predicted positives are actually positive.
    - It matters when the cost of a false positive is high.

<center>$$ Precision = \frac{TP}{TP+FP} $$</center>


- **Recall** (also called Sensitivity or True Positive Rate)
    - tells you what proportion of actual positives the model successfully caught.
    - It matters when missing a positive case is costly, such as in disease screening.

<center>$$ Recall = \frac{TP}{TP+FN} $$</center>

- **Specificity** (True Negative Rate) 
    - is the counterpart to recall for the negative class, measuring how well the model identifies actual negatives.
<center>$$ Specificity = \frac{TN}{TN+FP} $$</center>




- **F1 Score**
    - is the harmonic mean of precision and recall, giving a single balanced measure that penalizes extreme imbalance between the two.
 
      <center>$$ F1 = 2. \frac{Precision⋅Recall}{Precision+Recall} $$</center>


- **False Positive Rate (FPR)**
    - captures how often negatives are misclassified as positives

<center>$$ FPR = \frac{FP}{FP+TN} $$</center>

- **Area Under the ROC Curve (AUC)**
  
    - reliable for imbalanced data

    -  Extends the confusion matrix analysis beyond a single decision threshold
    -  AUC evaluates the model across all possible thresholds at once.
    -  It equals the probability that the model ranks a randomly chosen positive instance higher than a randomly chosen negative one. 

<center><center><img src="img/fig39.png" width=300 height=300 /></center></center>

- **Matthews Correlation Coefficient (MCC)**
    - reliable single-number summaries for imbalanced data

<center>$$ MCC = \frac{ (TP⋅TN)−(FP⋅FN)}{\sqrt{(TP+FP)(TP+FN)(TN+FP)(TN+FN)}} $$</center>

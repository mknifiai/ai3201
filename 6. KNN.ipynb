# Nearest Neighbor Methods

#### Learning
- Store all training examples
#### Prediction
- Find the training example $(x_i,y_i)$ such that $x_{i}$ is closest (for some notion of close) to $x'$
- Classify $x'$ with the label $y_{i}$

<center><img src="img/fig63.png" width=400 height=400 /></center>

- **For classification:** Every neighbor votes on the label. Predict the most frequent label among the neighbors.

- **For regression:** Predict the mean value

- k-nearest neighbors is an example of **instance-based learning methods**

**Nearest Neighbor Methods** are best for large data sets with only a few (< 20) attributes

**Advantages**
- Learning is easy

- Can learn very complex functions

**Disadvantages**
- Classification is slow $𝑂(dN)$ (need to keep the entire training set around)

    - K-Dimensional Trees can help simplify the classification task
    - by constructing a tree that partitions the search space
<space>
- Easily fooled by irrelevant attributes

### Distance between instances

- How do we measure distances between instances?

- **For numeric features:**

<center><img src="img/fig64.png" width=400 height=400 /></center>

<center> Euclidean = $L_{2}$ norm</center>

<center> Manhattan = $L{1}$ norm</center>


- **For symbolic/categorical features**
    - Most common distance is the Hamming distance
    - Number of bits that are different
    - Or: Number of features that have a different value
    - **Example:**
        - X1: {Shape=Triangle, Color=Red, Location=Left, Orientation=Up}
        - X2: {Shape=Triangle, Color=Blue, Location=Left, Orientation=Down}
        - Hamming distance = 2

### The Curse of Dimensionality

- Even if most/all features are relevant, in high dimensional spaces, most points are equally far from each other!
- Distances lose meaning
- Neighbors aren't similar

### Dealing with the curse of dimensionality

- Different ways of capturing the underlying dimensionality of the space
    - Eg: Dimensionality reduction techniques, manifold learning
<space>
- Feature selection

    - Different methods exist

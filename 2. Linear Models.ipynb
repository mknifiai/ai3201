### 1. Linear regression

**Input:** $(x_{1},y_{1}),\dots,(x_{n},y_{n})$

- $x_{n} \subseteq R^{d}$

**Goal:** find a function $f$ such that $f(x_{n})$ is a good approximation to $y_{n}$

**Hypothesis space:** a set of allowable functions $f: X \rightarrow Y$

- How do we measure the quality of $f$ during training. We'll use a **loss function** (error metric)

### Simple Linear regression

**Input** pairs of points $(x_{1},y_{1}),\dots,(x_{n},y_{n})$ with $x_{n} \in R$ and $y_{n} \in R$

**Hypothesis space** a set of linear functions $f(x) = ax+b$ with $a,b \in R$. 

i.e., $y_{n} = ax_{n}+b$

<center><img src="img/fig23.png" width=300 height=300 /></center>


**Why is the bias term needed?**

- to shift the line up or down so it does not have to pass through the origin 

**Loss function:** squared difference between the predicted value and the actual value

<center>$\frac{1}{2}(ax_{n} + b - y_{n})^{2}$</center>

<center><img src="img/fig24.png" width=150 height=150 /></center>


- In typical regression applications, measure the fit using **squared loss function**

<center>$L(h) = \frac{1}{N} \sum\limits_{i=1}^N \frac{1}{2}(f(x_{i}) - y_{i})^{2}$</center>


- We want to minimize the average loss on **the training data**

- The optimal linear hypothesis is given by

<center>$\min\limits_{a,b}\frac{1}{N} \sum\limits_{i=1}^N \frac{1}{2}(ax_{i}+b - y_{i})^{2}$</center>



**How do we find the optimal $a$ and $b$?**

- Solution 1: take the derviate and solve (there is a closed form solution)
    - $a = (XX^{T})^{-1}XY$ 
- Solution 2: use gradient descent or stochastic gradient descent
    - This apporach is more useful for general loss functions 

### 2. Gradient Descent

Iterative method to minimize a (convex) differentiable function $f$

- Find a direction along which the function is decreasing and
step in that direction

#### Convexity

<img src="img/fig25.png" width=500 height=500 />


<img src="img/fig26.png" width=500 height=500 />


**Iterative method to minimize a (convex) differentiable function $f$**

- Pick an intital point $x^{0}$
- Iterate until convergence
<center>$x^{t+1} = x^{t} - \lambda_{t} $$\nabla f(x^{t}) $</center>

where $\lambda_{t}$ is the $t^{th}$ step size (learning rate)

<img src="img/fig27.png" width=500 height=500 />

**Large learning rate:** Fast progress, but risks overshooting or diverging.

**Small learning rate:** Stable and reliable, but slow to converge.

### What if we enlarge the hypothesis class

- **Quadratic function** $\space$ $ax^{2} + bx + c$

- **k-degree polynomials** $\space$ $a_{k}x^{k} + a_{k-1}x^{k-1} + ax +a_{0}$

- could transform non-linearly separable data into linearly separable

- Larger hypothesis space typically decreases the cost function,
- but this does NOT necessarily mean better predictive performance
    - This phenomenon is known as **overfitting** (What is underfitting?)
    - Ideally, we would select the simplest hypothesis consistent with the observed data

### 3. Regression evaluation metrics

- **Mean Absolute Error**

<center><img src="img/fig29.png" width=200 height=200 /></center>

- Measures the average absolute distance between predictions and actual values
- robust to outliers.

- **Mean Squared Error (MSE)**

<center><img src="img/fig30.png" width=200 height=200 /></center>

- Measures the average squared difference
- heavily penalizes larger errors.

- **R-Squared**

<center><img src="img/fig31.png" width=400 height=300 /></center>

>> It quantifies the degree to which the variance in the dependent variable is predictable from the independent variables.

### 4. Train-Validation-Test split

- In practice, we can not simply evaluate our learned hypothesis on the training data
- we want it to perform well on unseen data 

We should split the dataset into three sets:

- **training set** apply the learning algorithms with different hyper-paprameters to produce $H = \{h_{1},\dots, h_{r}\}$
- **validation set** Choose $h_{i}$ from $H$ based on the validation set
- **test set** Estimiate the error of $h_{i}$ using test set

- The Train-Validation-Test is the best approach when the data is plentiful
- If data is scarce, we can use **K-fold cross validation**

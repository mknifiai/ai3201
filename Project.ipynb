## Comparative Analysis of Machine Learning Algorithms



## Overview

In this project, you will conduct a systematic comparison of four machine learning algorithms applied to a single dataset. 

The goal is to develop practical experience with the complete machine learning workflow, including data preparation, hyperparameter tuning, model evaluation, and critical comparison of algorithm performance. 

You may choose to focus on either a classification or a regression task, but your choice of dataset and algorithms must be consistent throughout.



## Dataset

You will work with a dataset of your choosing that is suitable for the task you select. The dataset should contain a sufficient number of samples and features to make the comparison meaningful. 

Popular sources include [the UCI Machine Learning Repository](https://archive.ics.uci.edu/), Kaggle, and scikit-learn's built-in datasets. 

Your chosen dataset must have a well-defined target variable appropriate for either classification or regression.

## Requirements

**1. Data Splitting**

You must partition your dataset into three distinct subsets: a training set, a validation set, and a test set. If your dataset does not come with a predefined split, you are responsible for creating one yourself. A common convention is a 60/20/20 or 70/15/15 ratio, but you should justify your choice. The test set must remain untouched until the final evaluation stage and must never be used during training or hyperparameter tuning.

**2. Feature Scaling (Normalization or Standardization)**

you must scale your features if they are not already scaled. You might use scikit-learn's StandardScaler or MinMaxScaler. Fit the scaler on the training set only, then use it to transform the validation and test sets to avoid data leakage. Briefly state which method you used in your report.

**3. Algorithm Selection**

Select four different algorithms appropriate for your chosen task. For example, a classification project might compare Logistic Regression, a Decision Tree, a Random Forest, and a Support Vector Machine, while a regression project might compare Linear Regression, k-Nearest Neighbors, Gradient Boosting, and a Neural Network. The algorithms should be reasonably distinct in their underlying assumptions so that the comparison is informative.

- Each of the four algorithms must be implemented in its own separate Python file (for example, logistic_regression.py, random_forest.py, and so on)
- Within each file, your code must be organized in one of two acceptable styles, and you should be consistent across all four files:

    - Pure functions
    - Class-based methods

**4. Hyperparameter Tuning**


For each of the four algorithms, you must identify the relevant hyperparameters and tune them using the validation set (or cross-validation restricted to the training portion). You should clearly document the range of hyperparameter values you explored, the search strategy you used (e.g., grid search or random search), and the final selected values for each model.

**5. Final Evaluation on the Test Set**

After all hyperparameters have been finalized, evaluate each of the four tuned models on the test set.

For classification tasks, you are required to report the confusion matrix for every algorithm on the test set, along with derived metrics such as accuracy, precision, recall, and F1-score.


For regression tasks, since a confusion matrix is not directly applicable, you must instead report appropriate error metrics such as Mean Squared Error (MSE), Root Mean Squared Error (RMSE), Mean Absolute Error (MAE), and the coefficient of determination($R^{2})$. 

## Deliverables
Submit a written report (approximately 6–10 pages) together with your complete, reproducible source code. 

Your report should include the following sections:

- An introduction describing the dataset, the task, and the motivation for your work.
- A methodology section detailing your data splitting strategy, the four algorithms, and your hyperparameter tuning procedure.
- A results section presenting the confusion matrices (for classification) or error metrics (for regression) for all four algorithms on the test set, supported by tables and figures.
- A discussion section that critically compares the algorithms, explains observed performance differences with reference to each method's assumptions and complexity, and reflects on any limitations.
- Finally, a conclusion summarizing which algorithm performed best for your problem and why.
